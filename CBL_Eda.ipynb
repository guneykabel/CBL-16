```python


crimes_per_region = df_2023.groupby(['Month', 'Falls within']).size().unstack()

plt.figure(figsize=(15, 6))
for column in crimes_per_region.columns:
    plt.plot(crimes_per_region.index, crimes_per_region[column], marker='o', linewidth=2, label=column)

plt.title('Crimes per Month by Region 2023')
plt.xlabel('Month')
plt.ylabel('Number of Crimes')
plt.xticks(rotation=45)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', fontsize=7)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

crimes_per_region = df_2023.groupby(['Month', 'Falls within']).size().unstack()

# Split regions into two halves
regions = crimes_per_region.columns.tolist()
half = len(regions) // 2
first_half = regions[:half]
second_half = regions[half:]

# Plot first half
plt.figure(figsize=(15, 6))
for col in first_half:
    plt.plot(crimes_per_region.index, crimes_per_region[col], marker='o', linewidth=2, label=col)
plt.title('Crimes per Month by Region 2023 (Part 1)')
plt.xlabel('Month')
plt.ylabel('Number of Crimes')
plt.xticks(rotation=45)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', fontsize=7)
plt.grid(True)
plt.tight_layout()
plt.show()

# Plot second half
plt.figure(figsize=(15, 6))
for col in second_half:
    plt.plot(crimes_per_region.index, crimes_per_region[col], marker='o', linewidth=2, label=col)
plt.title('Crimes per Month by Region 2023 (Part 2)')
plt.xlabel('Month')
plt.ylabel('Number of Crimes')
plt.xticks(rotation=45)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', fontsize=7)
plt.grid(True)
plt.tight_layout()
plt.show()
```
